#Visualizacion de datos con python: Matprlotlib, Seaborn y Plotly

```python
import pandas as pd
df = pd.read_csv("/content/canadian_immegration_data.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.set_index("Country", inplace=True)
df
```

```python
anos = list(map(str,range(1980,2014)))
anos
```

```python
colombia = df.loc["Colombia", anos]
colombia
```

```python
col_dict = {"Año":colombia.index.tolist(),
            "Inmigrantes":colombia.values.tolist()}
col_dict
```

```python
datos_col = pd.DataFrame(col_dict)
datos_col.tail()
```

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
plt.plot(datos_col["Año"],datos_col["Inmigrantes"])
plt.xticks(["1980", "1985", "1990", "1995", "2000", "2005", "2010"])
plt.show()
```

```python

plt.figure(figsize=(8,4))
plt.plot(datos_col["Año"],datos_col["Inmigrantes"])
plt.xticks(["1980", "1985", "1990", "1995", "2000", "2005", "2010"])
plt.title("Inmigracion de Colombianos hacia Canadá")
plt.xlabel("Año")
plt.ylabel("Número de Inmigrantes")
plt.show()
```

A partir de la gráfica se puede realizar la siguiente interpretación correcta y clara:

Entre 1980 y 1996, el número de inmigrantes colombianos hacia Canadá se mantuvo relativamente bajo y estable, siempre por debajo de los 2.000 inmigrantes anuales, con ligeras fluctuaciones pero sin cambios significativos en la tendencia general.

A partir de 1997, se observa un incremento sostenido y marcado en el flujo migratorio. Este crecimiento se mantiene de forma casi continua durante aproximadamente ocho años, lo que indica un aumento constante del interés o la necesidad de emigrar hacia Canadá durante ese período.

El flujo migratorio alcanza su punto máximo alrededor de 2005, con una cifra superior a los 6.000 inmigrantes, lo que representa el pico más alto de toda la serie analizada.

Posteriormente, desde 2006 en adelante, la gráfica muestra una tendencia descendente, con una disminución gradual en el número de inmigrantes, aunque los valores se mantienen más altos que los registrados antes de 1997. Esto sugiere que, pese a la caída, el nivel de inmigración no regresó a los mínimos históricos de las décadas anteriores.

En conjunto, la gráfica evidencia tres etapas claras: una fase de estabilidad baja, una fase de crecimiento acelerado y una fase de descenso moderado.

```python
fig, axs = plt.subplots(1,2, figsize=(10,4))
axs[0].plot(datos_col["Año"],datos_col["Inmigrantes"])
axs[0].set_title("Inmigracion de Colombianos \nhacia Canada 1980-2013")
axs[0].set_xlabel("Año")
axs[0].set_ylabel("Número de Inmigrantes")
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()

axs[1].boxplot(datos_col["Inmigrantes"])
axs[1].set_title("Bloxplit de Inmigrantes de Colombia \nhacia Canada 1980 - 2013")
axs[1].set_xlabel("Colombia")
axs[1].set_ylabel("Número de Inmigrantes")
axs[1].grid()

plt.show()
```

```python
fig, axs = plt.subplots(2,2, figsize=(10,6))
fig.subplots_adjust(hspace=0.5,wspace=0.3 )
fig.suptitle("Inmigracion de los 4 mayores paises sudamericanos \nhacia Canada desde 1980 hasta 2013")

axs[0,0].plot(df.loc["Colombia",anos])
axs[0,0].set_title("Colombia")

axs[0,1].plot(df.loc["Brazil",anos])
axs[0,1].set_title("Brazil")

axs[1,0].plot(df.loc["Argentina",anos])
axs[1,0].set_title("Argentina")

axs[1,1].plot(df.loc["Peru",anos])
axs[1,1].set_title("Peru")

ymin = 0
ymax = 7000

for ax in axs.ravel():
  ax.set_ylim(ymin,ymax)


for ax in axs.flat:
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))
  ax.set(xlabel="Año", ylabel="Número de Inmigrantes")
  ax.grid()

plt.show()
```

```python
fig, ax =plt.subplots(figsize=(8,4))
ax.plot(datos_col["Año"], datos_col["Inmigrantes"],lw=3, color="g")
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title("Inmigrantes de Colombia hacia Canada \n 1980-2013", fontsize=18, loc="left")
ax.set_xlabel("Año", fontsize=14)
ax.set_ylabel("Número de inmigrantes", fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
ax.grid()
plt.show()
```

```python
sudamerica = df.query("Region == 'South America'")
sudamerica.head(10)
```

```python
colores = ["royalblue", "orange", "forestgreen", "orchid", "purple", "brown", "slateblue", "gray", "olive","navy", "teal", "tomato"]

fig, ax = plt.subplots(figsize=(12,5))
ax.bar(sudamerica.index, sudamerica["Total"], color=colores)
ax.set_title("inmigrantes de sudamenicanos hacia \n Canada durante el perido de 1980 hasta 2013",
loc="left", fontsize=18)
ax.set_xlabel(" ")
ax.set_ylabel("Número de Inmigrantes", fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

plt.show()

```

```python
colores = ["royalblue", "orange", "forestgreen", "orchid", "purple", "brown", "slateblue", "gray", "olive","navy", "teal", "tomato"]

fig, ax = plt.subplots(figsize=(12,5))
ax.barh(sudamerica.index, sudamerica["Total"], color=colores)
ax.set_title("inmigrantes de sudamenicanos hacia \n Canada durante el perido de 1980 hasta 2013",
loc="left", fontsize=18)
ax.set_ylabel(" ")
ax.set_xlabel("Número de Inmigrantes", fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

plt.show()
```

```python
sudamerica_sorted = sudamerica.sort_values("Total", ascending=True)
colores = ["royalblue", "orange", "forestgreen", "orchid", "purple", "brown", "slateblue", "gray", "olive","navy", "teal", "tomato"]

fig, ax = plt.subplots(figsize=(12,5))
ax.barh(sudamerica_sorted.index, sudamerica_sorted["Total"], color=colores)
ax.set_title("inmigrantes de sudamenicanos hacia \n Canada durante el perido de 1980 hasta 2013",
loc="left", fontsize=18)
ax.set_ylabel(" ")
ax.set_xlabel("Número de Inmigrantes", fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

for i,j in enumerate(sudamerica_sorted["Total"]):
  ax.text(j+20 , i, str(j), color= "black", fontsize=8.5, va="center")
  plt.show()
```

```python
colores =[]
for pais in sudamerica_sorted.index:
  if pais == "Colombia":
    colores.append("forestgreen")
  else:
    colores.append("silver")

fig, ax = plt.subplots(figsize=(12,5))

ax.set_frame_on(False)
ax.get_xaxis().set_visible(False)
ax.tick_params(axis="y", length=0)

ax.barh(sudamerica_sorted.index, sudamerica_sorted["Total"], color=colores)
ax.set_title("Subamérica: Colombia es el segundo pais con más migrantes \n hacia Canada durante el perido de 1980 hasta 2013",
loc="left", fontsize=18)
ax.set_ylabel(" ")
ax.set_xlabel("Número de Inmigrantes", fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=10)


for i,j in enumerate(sudamerica_sorted["Total"]):
  ax.text(j+20 , i, str(j), color= "black", fontsize=8.5, va="center")


plt.show()


```
